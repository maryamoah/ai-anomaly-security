# Interpretable Anomaly Detection for Security Telemetry

This notebook demonstrates a lightweight anomaly detection pipeline using Isolation Forest.

```python
import sys
sys.path.append("../src")

from preprocess import load_data, prepare_features
from model import train_isolation_forest, score_anomalies
from evaluate import plot_anomaly_distribution

import pandas as pd
print("Imports successful")
```
